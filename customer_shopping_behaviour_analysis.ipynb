```python
import pandas as pd 
import numpy as np 
```

```python
df = pd.read_csv("customer_shopping_behavior.csv")
df.head(5)
# df.info()
df.describe() 
```

```python
df.dtypes
```

```python
df.columns = df.columns.str.replace(" ","_ ")
df.head()
```

```python
df.rename( columns = {"Purchased_Amount_(USD)" : "purchased_amount"})
```

```python
df.columns = df.columns.str.lower() 
df.head()
```

```python
age_cuts = ["Young-Adult","Adult","Middle-aged","Senior"]
df['age_criteria'] = pd.qcut(df['age'], q=4,labels = age_cuts) 
df.columns
```

```python
days_assign = {
     'Fortnightly' : 14, 'Weekly' : 7, 'Annually':365, 'Quarterly':90, 'Bi-Weekly' : 14,
       'Monthly' : 30, 'Every 3 Months' : 90
 }
df['freq_of_purchase_in_days'] =  df['frequency_of_purchases'].map(days_assign)
```

```python
df.head()
```

```python
df.rename( columns = { 'purchase_amount_(usd)' : 'purchase_amount' } , inplace=True )
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df.drop( columns = ('promo_code_used') , inplace = True)
```
